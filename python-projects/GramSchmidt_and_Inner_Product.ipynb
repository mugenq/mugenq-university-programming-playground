# Solution

Part 1: Check if (R^3,(-,-)_A) is a metric vector space
# Properties to be satisfied:
A metric vector space is a vector space with a metric that measures the distance between two vectors. The metric should satisfy the following properties:

    Non-negativity: d(x,y) ≥ 0 for all x, y and d(x,y) = 0 if and only if x = y.
    Symmetry: d(x,y) = d(y,x) for all x, y.
    Triangle inequality: d(x,y) + d(y,z) ≥ d(x,z) for all x, y, z.

Here, we are given a matrix A and the metric (x,y)_A = Transpose(Ax).Ay. We need to check if it satisfies the above properties.

Part 2: Find the orthonormal basis using the Gram-Schmidt orthonormalization method
# The Gram-Schmidt process is a method for orthonormalizing a set of vectors in an inner product space, most commonly the Euclidean space R^n.

# Given a linearly independent set of vectors {v1, v2, v3}, the process is as follows:

    Set u1 = v1.
    For j = 2,3,...,n, set uj = vj - sum((vj,ui)/(ui,ui) * ui) for all i less than j.
    Normalize each vector uj to get the orthonormal basis {e1, e2, e3}, where ej = uj/||uj||.

Here, the given vectors v1 = (1,1,-1), v2 = (-1,1,0), and v3 = (0,1,3). Applying the Gram-Schmidt process, we get:

    u1 = v1 = (1,1,-1)
    u2 = v2 - (v2,u1)/(u1,u1) * u1
    u3 = v3 - (v3,u1)/(u1,u1) * u1 - (v3,u2)/(u2,u2) * u2
    Normalize each uj to get ej.

You would need to plug in the values of v1, v2, and v3 in the above equations to get the exact values of u2, u3 and subsequently e1, e2, and e3 4 5.

Remember to use the given inner product (-,-)_A instead of the standard dot product when computing the projections in the Gram-Schmidt process.


# Here is the python code:

```python
# First of all, let us define the dot product function as stated in the question.

import numpy as np

def dotProd(x, y):
    """Compute the dot product using the specified matrix A."""
    A = np.array([[1, 1, 1], [0, 1, 1], [0, 0, 1]])
    return np.matmul(np.transpose(np.matmul(A, np.array(x))), np.matmul(A, np.array(y)))
```

```python
# From that, we will need to define a projecting function and a normalizing function based on the new dot product function.
# We can supply the dot function as the argument dot=dotProduct later on, otherwise use dot=np.dot for usual euclidian inner product.

def proj(u0, v0, dot=np.dot):
    """Compute the projection of vector v0 onto vector u0."""
    u, v = np.array(u0), np.array(v0)
    return dot(u, v) / dot(u, u) * u

def normal(v0, dot=dotProd):
    """Normalize the vector v0 using the specified dot product."""
    v = np.array(v0)
    return v / np.sqrt(dot(v, v))
```

```python
# Here we will get into the GramSchmidt basis process

def gramSchmidt(vList=[], dot=np.dot):
    """Perform Gram-Schmidt orthonormalization on a list of vectors."""
    basis = []
    vectorList = [np.array(v) for v in vList] # For the sake of ease of input, we can accept a list of python lists as arguments and convert to a numpy array matrix.
    
    for v in vectorList:
        # Orthogonalize each vector against the basis, in succession
        w = v - sum(proj(b, v, dot) for b in basis) # We use a generator to perform the summation (np.sum summation function has intricacies, hence normal sum)
        
        # Check if the resulting vector w is not too close to zero
        if np.any(w > 1e-10):
            # Normalize the orthogonalized vector and add to the basis
            basis.append(normal(w, dot))
    
    return np.array(basis)
```

```python
def checkOrthonormal(vList=[], dot=np.dot):
    """Check if the vectors in vList form an orthonormal basis."""
    vectorList = [np.array(v) for v in vList]
    AtA_matrix = np.array([[dot(a, b) for b in vectorList] for a in vectorList]) # Test for orthonormality being true <=> A^t * A = I
    return np.allclose(AtA_matrix, np.eye(len(vectorList)))
```

```python
# Example usage
gramArray = gramSchmidt([[1, 1, -1], [-1, 1, 0], [0, 1, 3]], dotProd)

check = [[dotProd(a, b) for b in gramArray] for a in gramArray]

# Output results and check if the basis is orthonormal
gramArray, check, checkOrthonormal(gramArray, dotProd)
```
