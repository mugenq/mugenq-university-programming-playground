Lazer DT Tourney:

Multiple rounds half-half elimination playlists
Example: 8 players -> cut bottom 4 scores - > 4 player playlist -> cut bottom 2 -> etc...

Scoring is Lazer V2, standardized. nomodSS = 1 million

All maps must be played with rate adjust. (1.01 - 2.00, floor is decided per map, ceil is always 2.00)
Scoring: Score * DT rate (*1.2 with HR or EZ or FL. if HRFL or EZFL, then 1.44 for the pair)
EZ halves the score, hence we normalize it to 1mil instead of 500k before score calculation. 
Because of the OD, HR offers 1.06 advantage over EZ.
Example: DTHR -> (Score after mods) * 1.5 * 1.2 = (Score before mods) * 1.7 * 1.5 * 1.2
Example: HDDT(2x) -> (Score before mods) * 1.28 * 2 = (Score before mods) 2.56

When choosing freestyle, multiply score with the star rating of the difficulty chosen. Scoring is decked for star ratings below the
main difficulty of the diffset. The decked amount if 10k per 0.01 star rating difference.
Example: Main 6.2, Selected 6.5 DT SS = 6.5 * 1mil * 1.1 * 1.5 = 10.725 mil
Example: Main 6.2, Selected 3.2 HDDTHR SS = 3.2 * 1mil * 1.24 * 1.5 * 1.2 - 3mil = 4142400

Other mods:
Traceable, blinds receive the scoring from Hidden and Flashlight, respectively
Muted mod has 1x score multiplier
Sudden death, perfect, accuracy challenge, strict tracking yield 1x score multiplier
Any other mod is not allowed.

Three star system:
First star: Bonus 500k Score by using a rate above the floor.
Second star: Bonus 500k Score for using a specified mod-set (allow to stack with other mods)
Third star: Bounty Hunter (if the map has any kind of bounty attached to it)

All multipliers are applied before the bonuses.

```python
def calc_score(mainSR, selectedSR, lazer_score, dtRate, hrBool, hdBool, ezBool, flBool, nfBool):
    # settings
    hrBonus = 0.2
    flBonus = 0.2
    hdBonus = 0
    ezBonus = 1.1
    nfBonus = 1
    scorePenalty = 1000000

    # calculate score detail
    score = selectedSR * lazer_score * dtRate

    score = score * (hrBool*hrBonus + 1) * (hdBool*hdBonus + 1) * (ezBool*ezBonus + 1) * (flBool*flBonus + 1) * (nfBool*nfBonus + 1)
    if mainSR - selectedSR > 0:
        score = score - (mainSR - selectedSR)*(scorePenalty) # penalize for selecting lower star rating 
    score = score / 10**6
    return max(score,0)
```

```python
# [HR HD EZ FL NF]
mainSR = 5
scoreMainDT = [mainSR, 5, 650000, 1.5, False, False, False, False, False]
scoreSub2X = [mainSR, 3, 1050000, 2, False, False, False, False, False]
scoreSubDTHR1 = [mainSR, 4, 800000, 1.5, True, False, False, False, False]
scoreSubDTHR2 = [mainSR, 3, 1100000, 1.7, True, False, False, False, False]
scoreEZ = [mainSR, 5, 300000, 1.5, False, False, True, False, False]
scoreShredder = [mainSR, 6, 250000, 1.5, False, False, False, False, True]

print(calc_score(*scoreMainDT))
print(calc_score(*scoreSub2X))
print(calc_score(*scoreSubDTHR1))
print(calc_score(*scoreSubDTHR2))
print(calc_score(*scoreEZ))
print(calc_score(*scoreShredder))
```

```python
import random

def random_until_214():
    found = False
    while not found:
        batch = []
        for _ in range(10):
            num = random.randint(1, 214)
            batch.append(num)
            if num == 214:
                found = True
                break
        print(*(batch))

random_until_214()
```
