```python
def perm(x =[], p=[]):
    permutationlist = []
    for item in p:
        permutationlist.append(x[item])
    return permutationlist
 
def compose(x, f= []):
    if len(f) == 0:
        return x
    else:
        return compose(perm(x, f[0]), f[1:])

def inv(x=[]):
    mapping_list = []
    for iterator in range(0, len(x)):
        mapping_list.append([x[iterator], iterator])
    mapping_list = [item[1] for item in sorted(mapping_list, key=lambda k: k[0])]
    return mapping_list 

def sgn(values):
    # Simple Shell sort; while O(N^2), we only deal with at most 4 values 
    values = list(values)
    N = len(values)
    num_swaps = 0
    for i in range(N-1):
        for j in range(i+1, N):
            if values[i] > values[j]:
                values[i], values[j] = values[j], values[i]
                num_swaps += 1
    return (-1)**(num_swaps % 2)

def to_cycles(perm):
    pi = {i+1: perm[i] for i in range(len(perm))}
    cycles = []

    while pi:
        elem0 = next(iter(pi)) # arbitrary starting element
        this_elem = pi[elem0]
        next_item = pi[this_elem]

        cycle = []
        while True:
            cycle.append(this_elem)
            del pi[this_elem]
            this_elem = next_item
            if next_item in pi:
                next_item = pi[next_item]
            else:
                break
        
        cycle.insert(0,cycle.pop(len(cycle)-1))

        if len(cycle) > 1:
            cycles.append(cycle)

    return cycles
```

```python
import operator as op
from functools import reduce

def ncr(n, r):
    r = min(r, n-r)
    numer = reduce(op.mul, range(n, n-r, -1), 1)
    denom = reduce(op.mul, range(1, r+1), 1)
    return numer // denom  # or / in Python 2

matrix = []
vec = []
size = 8
for i in range(1, size+1):
    for j in range(1, size+1):
        vec.append(ncr(i+j-2, i-1))
    matrix.append(list(vec))
    print(vec)
    vec.clear()

print(matrix)

import numpy as np

print(np.linalg.det(matrix))
```

```python
import numpy as np

def dotProd(x, y):
    """Compute the dot product using the specified matrix A."""
    A = np.array([[1, 1, 1], [0, 1, 1], [0, 0, 1]])
    return np.matmul(np.transpose(np.matmul(A, np.array(x))), np.matmul(A, np.array(y)))

def proj(u0, v0, dot=np.dot):
    """Compute the projection of vector v0 onto vector u0."""
    u, v = np.array(u0), np.array(v0)
    return dot(u, v) / dot(u, u) * u

def normal(v0, dot=dotProd):
    """Normalize the vector v0 using the specified dot product."""
    v = np.array(v0)
    return v / np.sqrt(dot(v, v))

def gramSchmidt(vList=[], dot=np.dot):
    """Perform Gram-Schmidt orthonormalization on a list of vectors."""
    basis = []
    vectorList = [np.array(v) for v in vList]
    
    for v in vectorList:
        # Orthogonalize each vector against the basis
        w = v - sum(proj(b, v, dot) for b in basis)
        
        # Check if the resulting vector w is not too close to zero
        if np.any(w > 1e-10):
            # Normalize the orthogonalized vector and add to the basis
            basis.append(normal(w, dot))
    
    return np.array(basis)

def checkOrthonormal(vList=[], dot=np.dot):
    """Check if the vectors in vList form an orthonormal basis."""
    vectorList = [np.array(v) for v in vList]
    AtA_matrix = np.array([[dot(a, b) for b in vectorList] for a in vectorList]) # Test for orthonormality being true <=> A^t * A = I
    return np.allclose(AtA_matrix, np.eye(len(vectorList)))

# Example usage
gramArray = gramSchmidt([[1, 1, -1], [-1, 1, 0], [0, 1, 3]], dotProd)

check = [[dotProd(a, b) for b in gramArray] for a in gramArray]

# Output results and check if the basis is orthonormal
gramArray, check, checkOrthonormal(gramArray, dotProd)
```

```python
def funcMod(input_var=0, input_cons=1):
    return input_var*input_cons

def findMod(mod, n, base, argFuncMod=funcMod):
    if n > 0:
        return argFuncMod(findMod(mod, n-1, base, argFuncMod), base)%mod
    else:
        return 1
```
