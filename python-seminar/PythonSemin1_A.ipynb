Defining Functions in Python:

Functions allow us to encapsulate reusable pieces of code. They improve code readability and maintainability.

```python
# Example of a function
def greet(name):
    return f"Hello, {name}!"

# Using the function
result = greet("Alice")
print(result)
```

In Python, you can apply a function to each element of a list using the map function or using a list comprehension. Here's an example of both approaches:

```python
def square(x):
    return x ** 2

original_list = [1, 2, 3, 4, 5]

# Using map to apply the square function to each element of the list
result = list(map(square, original_list))

print(result)
```

Introduction to Python Generators (LIST COMPREHENSION):

Generators are a memory-efficient way to iterate over large datasets. They produce values on-the-fly, saving memory compared to storing an entire list in memory.

On top of that, we can automatically generate lists with generators.

```python
"""Brief explanation of range(0,1): range is also a generator, takes 0 and 1 as input and
becomes an object."""

# Example of a generator function
def square_generator(n):
    for i in range(n):
        yield i**2 # this is similar to a "return" when defining a python function

# Using the generator
squares = square_generator(5)
print(list(squares)) # list() is a function as well. it takes inputs such as a generator and converts them

# Cleaner way of creating lists
square_list_clean = [i**2 for i in range(0,10)]
print(square_list_clean)
```

We can use generators for the previous map usage.

```python
def square(x):
    return x ** 2

original_list = [1, 2, 3, 4, 5]

# Using list comprehension to apply the square function to each element of the list
result = [square(x) for x in original_list]

print(result)
```

A throwaway function, often expressed using a lambda function, can be used in list comprehension for various purposes. Here's a simple example where we use a throwaway function to square each element of a list:

```python
# Original list
numbers = [1, 2, 3, 4, 5]

# Using a throwaway lambda function in list comprehension to square each element
squared_numbers = [(lambda x: x**2)(num) for num in numbers]

# Display the result
print(squared_numbers)
```
