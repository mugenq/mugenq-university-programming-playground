Let us recap about creating numpy arrays, and manipulating them. check the following code. As a handle for MPL, numpy is essential.

```python
mat = np.array([[1,1],[1,2]])

def ahh(a, n, fun):
  if n == 0:
    return a
  else:
    return fun(ahh(a, n-1, fun), a)

print(ahh(mat, 7, np.dot))
```

Format Strings fmt

You can also use the shortcut string notation parameter to specify the marker.

This parameter is also called fmt, and is written with this syntax:

**Back to Python List comprehension, have a look on the next few lines of code:**

Conditionals in list comprehension are also useful. Let us take a look:

```python
value = 123
print(value, 'is', 'even' if value % 2 == 0 else 'odd') # Conditional used
```

```python
# choose between f or g [f(x) if condition else g(x) for x in sequence]

mod_7_square_table = [True if i**2%7==1 else False for i in range(0,10)]
print(mod_7_square_table)
```

```python
# include only if [f(x) for x in sequence if condition]

mod_7_square = [i**2 for i in range(0,100) if i**2%7==1]
print(mod_7_square)
```

```python
# chooose between [f condition1 g] only if condition2 [f(x) if condition1 else g(x) for x in sequence if condition2]

mod_7_square_even = ["Even" if i**2%2==0 else "Odd" for i in range(0,100) if i**2%7==1]
print(mod_7_square_even)
```

**Histogram are also possible with MPL. We will use them a lot when talking about normal distributions.**

Not only that, we will also take a look on *Object Oriented* type of MPL coding.

Search the internet for an explanation of what **Object Oriented Programming** is.

```python
# Generate some example data
s = np.random.normal(0, 1, 200)  # Replace this with your actual data

"""Plotting the Histogram"""

# Create a figure and axis objects
fig, ax = plt.subplots()

# Plot the histogram (we need to define the qualities of the bar)
count, bins, ignored = ax.hist(s, 50, density=True) ## count is the height of the bar // bins is the number of bars // ignored represents any data that falls outside of the bars

"""Plotting the Normal Distribution"""

# Plot the probability density function (PDF)
mu, sigma = np.mean(s), np.std(s)

def N(x, mu=0, sigma=1):
  return 1 / (sigma * np.sqrt(2 * np.pi)) * np.exp(- (x - mu)**2 / (2 * sigma**2))

pdf = N(bins, mu, sigma)
ax.plot(bins, pdf, linewidth=2, color='r')

# Set labels for the axes
ax.set_xlabel("Value")
ax.set_ylabel("Probability Density")

# Show the plot
plt.show()

"""Try printing some of the variables used in the code. It is important to visualize what we are doing.
Since we are also using Object Oriented Programming, it is important to have a good grasp on the basics, and practice."""
```
