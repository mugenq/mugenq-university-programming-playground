Let us apply these in practice with a coin tossing problem!

```python
import random

def simulate_coin_tosses(num_sets, tosses_per_set):
    results_dict = {}

    for _ in range(num_sets):
        tosses = [random.choice(['Heads', 'Tails']) for _ in range(tosses_per_set)]
        total_heads = tosses.count('Heads')

        results_dict[total_heads] = results_dict.get(total_heads, 0) + 1

    return results_dict

# Simulate 1000 sets of 5 coin tosses
num_sets = 1000
tosses_per_set = 5
results = simulate_coin_tosses(num_sets, tosses_per_set)
sorted_dict_by_keys = {k: results[k] for k in sorted(results)}

# Print the results
print(sorted_dict_by_keys)
```

Onto Statistics:

Let us assume that we have a distribution X from a population P_pop. The follow definitions are as as headed:

- X ∼ Y, when X and Y are equivalent distributions
- X ∼ B(n, p) or X ∼ Bin(n, p) when X follows a binary distribution. Example: X = {0:1, 1:4, 2:6, 3:4, 4:1}
- X ∼ N(n, σ²) when X follows a normal distribution.

Equivalent Distributions (X ∼ Y):

    When we say X ∼ Y, it means that random variables XX and YY follow equivalent probability distributions. In other words, they have the same probability mass function (PMF) or probability density function (PDF), depending on whether they are discrete or continuous distributions.

In other words, two probability distributions being equal means that the two distributions have the same probabilities for every outcome in the sample space. This means that for any given event, the probability of that event occurring under both distributions is the same.

Mathematically, if X and Y are two random variables with probability distributions p(x) and q(x) respectively, then X and Y are equal if and only if p(x) = q(x) for all values of x in the sample space. This means that for any value of x, the probability that X takes the value x is the same as the probability that Y takes the value x.

In simpler terms, two probability distributions are equal if they have the same likelihood of assigning a probability to each possible outcome in the sample space.

Binomial Distribution (X ∼ B(n, p) or X ∼ Bin(n, p)):

A binomial distribution models the number of successes in a fixed number of independent and identically distributed binary trials (experiments), where each trial has two possible outcomes: success (usually denoted as 1) or failure (usually denoted as 0). The parameters nn and pp represent the number of trials and the probability of success, respectively.

Example Code:

```python
import random

# Parameters for the binomial distribution
n = 4  # Number of trials
p = 0.5  # Probability of success

# Generating a binomial distribution
X_values = [sum(random.choices([0, 1], weights=[1-p, p], k=n)) for _ in range(1000)]

# Counting occurrences using dictionary and list comprehension
X_counter = {value: X_values.count(value) / 1000 for value in set(X_values)}

# Displaying the binomial distribution
for value, probability in sorted(X_counter.items()):
    print(f"Number of Successes: {value}, Probability: {probability}")
```

Normal Distribution (X ∼ N(μ, σ²)):

A normal distribution, also known as a Gaussian distribution, is a continuous probability distribution that is symmetric around its mean (μμ). The parameter σσ represents the standard deviation, which indicates the spread or variability of the distribution.

Example Code:

```python
import random
import math

# Parameters for the normal distribution
mu = 0  # Mean
sigma = 1  # Standard deviation

# Generating a normal distribution
X_values = [mu + sigma * math.sqrt(-2 * math.log(random.random())) * math.cos(2 * math.pi * random.random()) for _ in range(10)]

# Displaying the normal distribution
for value in X_values:
    print(f"X: {value}")
```
