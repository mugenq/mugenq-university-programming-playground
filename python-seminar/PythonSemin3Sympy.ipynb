```python
import sympy as sy
from IPython.display import display, Math

def compute_first_order_partial_derivatives(f, variables=("x", "y")):
    return sy.simplify(sy.diff(f, x)), sy.simplify(sy.diff(f, y))

def compute_second_order_partial_derivatives(f, variables=('x', 'y')):
    x, y = sy.symbols(variables)
    
    # Compute second-order partial derivatives
    f_xx = sy.simplify(sy.diff(f, x, x))
    f_xy = sy.simplify(sy.diff(f, x, y))
    f_yx = sy.simplify(sy.diff(f, y, x))
    f_yy = sy.simplify(sy.diff(f, y, y))
    
    return f_xx, f_xy, f_yx, f_yy

def find_tangent_plane_expression(f, a, b):
    x, y = sy.symbols('x y')
    
    # Compute function value and partial derivatives at (a, b)
    f_value = f.subs({x: a, y: b})
    df_dx = sy.diff(f, x).subs({x: a, y: b})
    df_dy = sy.diff(f, y).subs({x: a, y: b})
    
    # Expression for the tangent plane
    tangent_plane_expr = f_value + df_dx * (x - a) + df_dy * (y - b)
    
    return tangent_plane_expr

def taylor_series_2nd_degree(f, a, b):
    x, y = sy.symbols('x y')

    # Compute first-order partial derivatives
    fx = sy.diff(f, x)
    fy = sy.diff(f, y)

    # Compute second-order partial derivatives
    fxx = sy.diff(fx, x)
    fxy = sy.diff(fx, y)
    fyy = sy.diff(fy, y)

    # Define the tangent plane equation
    tangent_plane = f.subs({x: a, y: b}) + fx.subs({x: a, y: b})*(x - a) + fy.subs({x: a, y: b})*(y - b)
    # Compute the quadratic terms
    quadratic_terms = sy.Rational(1/2) * (fxx.subs({x: a, y: b})*(x - a)**2 + 2*fxy.subs({x: a, y: b})*(x - a)*(y - b) + fyy.subs({x: a, y: b})*(y - b)**2)
    # Construct the expression for the second-degree Taylor series expansion
    taylor_series = tangent_plane + quadratic_terms

    return sy.simplify(taylor_series)

# Example usage:
# Define your function f(x, y)
x, y = sy.symbols('x y')
a, b = x, y
f = x**2 * y**2 - x + y


display(Math('f(x, y) = ' + sy.latex(f))) 

for expr, label in zip(compute_first_order_partial_derivatives(f), ['f_{x}', 'f_{y}']):
    display(Math(f'{label} = ' + sy.latex(expr.subs({x:a, y:b}))))

for expr, label in zip(compute_second_order_partial_derivatives(f), ['f_{xx}', 'f_{xy}', 'f_{yx}', 'f_{yy}']):
    display(Math(f'{label} = ' + sy.latex(expr.subs({x:a, y:b}))))
```

```python
import sympy as sy

def Lag_Crit(f, g):
    # Extract all symbols from the expressions f and g
    all_symbols = set().union(f.free_symbols, g.free_symbols)

    # Define variables
    variables = list(all_symbols)
    λ = sy.symbols('λ')
    # Define the objective function and constraint function
    F, G= f, g
    # Define the Lagrangian
    L = F - λ * G

    # Calculate partial derivatives
    derivatives = [sy.diff(L, var) for var in variables]
    derivatives.append(G)  # Add the constraint to the system

    # Solve the system of equations
    solution = sy.solve(derivatives, variables + [λ], dict=True)

    # Extract critical points and corresponding function values
    critical_points = [[sol[var] for var in variables] for sol in solution]
    f_values = [F.subs({var: sol[var] for var in variables}) for sol in solution]

    # Find max and min
    f_max, f_min = max(f_values), min(f_values)

    return critical_points, f_max, f_min

# Example usage:
# Define your objective function and constraint
x, y = sy.symbols('x y')
f = x*y  # Replace with your objective function
g = x**2 + x*y + y**2 - 1  # Replace with your constraint

# Call the function
critical, f_max, f_min = Lag_Crit(f, g)

# Print the results
print("Critical Points:", critical)
print("Maximum Value of f:", f_max)
print("Minimum Value of f:", f_min)
```
