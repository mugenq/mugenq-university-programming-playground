1.1 Sets work like normal mathematical sets. You can think of them the same as a list, but without repetitions.

```python
myset = {1, 2, 3}

myset.add(5)
myset.update({4,6})
myset.discard(2)

print(myset)
if 2 in myset:
    print(2)
else:
    print(f"not {2}")

newset = {1, 3, 5, 7}

print(myset.union(newset), myset.intersection(newset))
```

1.2 Dictionaries work like sets, though they have the added functionality of retrieving values, given a certain key.

```python
def calculate_average(grades):
    return sum(grades.values()) / len(grades)

# Example usage:
student_grades = {'Math': 90, 'English': 85, 'Science': 92}
average_score = calculate_average(student_grades)
print(f"Average Score: {average_score}")

print(student_grades["Math"])
```

```python
keys = ["name", "age", "city"]
values = ["Alice", 30, "Wonderland"]
person_dict = {key: value for key, value in zip(keys, values)}
print(f"Person Dictionary: {person_dict}")
```

2.1 Next, we can make use of our current folder as a workspace for our python programs. Acessing files and data sheets. Hence, we are able open, write to files. Not only that, we can also download stuff from the internet with the appropriate libraries.

```python
def write_to_file(students, filename='student_data.txt'):
    with open(filename, 'w') as file:
        for student in students:
            file.write(f"Name: {student['name']}, Age: {student['age']}, Country: {student['country']}\n")

# Example usage:
student_list = [{'name': 'Alice', 'age': 20, 'country': 'USA'},
                {'name': 'Bob', 'age': 22, 'country': 'Canada'}]
write_to_file(student_list)
```

3. Next and perhaps the most interesting is the class functionality. We can define whole structures and behaviours around them together with their own functions.

They are very useful for specific cases, for example creating objects for physics simulation, or aggregates of data inside an object.

In this case you can see that the class is used to define a class of complex numbers. We can also use classes to define structures such as polynomials.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")

# Creating an instance of the class
person1 = Person("John Doe", 25)

# Accessing attributes and calling a method
print(person1.name)  # Accessing the 'name' attribute
print(person1.age)   # Accessing the 'age' attribute
person1.display_info()  # Calling the 'display_info' method
```

```python
class Complex:
    def __init__(self, real, imag): ## initialize
        self.real = real
        self.imag = imag

    ## ADD
    def __add__(self, other):
        if isinstance(other, (int, float)):
            other = Complex(other, 0)
        return Complex(self.real + other.real, self.imag + other.imag)
    def __radd__(self, other): ## right side addition (remove and its unsupported)
         return self + other

    ## SUBTRACT
    def __sub__(self, other):
        if isinstance(other, (int, float)):
            other = Complex(other, 0)
        return Complex(self.real - other.real, self.imag - other.imag)
    def __rsub__(self, other):
         return self - other

    ## MULTIPLICATE
    def __mul__(self, other):
        if isinstance(other, (int, float)):
            other = Complex(other, 0)
        real_part = (self.real * other.real) - (self.imag * other.imag)
        imag_part = (self.real * other.imag) + (self.imag * other.real)
        return Complex(real_part, imag_part)
    def __rmul__(self, other):
         return self * other

    def __repr__(self): ## represent (when you put the variable alone)
        return f"{self.real} + {self.imag}j"
    def __str__(self): ## string (you can call str() on the function or even print() without having python print "an object at memory location")
        return f"{str(self.real)} + {str(self.imag)}j"
    
    @staticmethod ## a decorator for defining a function that belongs to the class, rather to an object of that class. that is why it does not take "self" as input
    def i(n=1):
        return Complex(0, n)


# Creating instances of the ComplexNumber class
complex1 = Complex(2, 3)
z = 1 + 2*Complex.i()  # Using the Complex.i method to create a complex number

print(z)
print((1 + Complex.i()) * (1 - Complex.i()))
```
